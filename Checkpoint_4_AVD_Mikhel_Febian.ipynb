## **Pengecekan Outliers**

```python
results = []

cols = baca_dataset.select_dtypes(include=['float64', 'int64'])

for col in cols:
  q1 = baca_dataset[col].quantile(0.25)
  q3 = baca_dataset[col].quantile(0.75)
  iqr = q3 - q1
  lower_bound = q1 - 1.5*iqr
  upper_bound = q3 + 1.5*iqr
  outliers = baca_dataset[(baca_dataset[col] < lower_bound) | (baca_dataset[col] > upper_bound)]
  percent_outliers = (len(outliers)/len(baca_dataset))*100
  results.append({'Kolom': col, 'Persentase Outliers': percent_outliers})

# Dataframe dari list hasil
results_df = pd.DataFrame(results)
results_df.set_index('Kolom', inplace=True)
results_df = results_df.rename_axis(None, axis=0).rename_axis('Kolom', axis=1)

# Tampilkan dataframe
display(results_df)
```

Dari pengecekan outliers diatas ada beberapa kolom yang memiliki outliers yaitu:

2. reading score	sebanyak 0.6%
3. writing score	sebanyak 0.5%

# **Construct Data**

```python
baca_dataset['average_score'] = baca_dataset[['math score', 'reading score', 'writing score']].mean(axis=1)
def klasifikasi_performa(score):
    if score >= 80:
        return 'High'
    elif score >= 60:
        return 'Medium'
    else:
        return 'Low'

baca_dataset['performance_category'] = baca_dataset['average_score'].apply(klasifikasi_performa)

display(baca_dataset[['average_score', 'performance_category']].head())
```
